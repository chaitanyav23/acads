# Classes
- A core concept of _object-oriented programming_
- Think of classes like datatypes -- all Python datatypes are indeed classes
    - `int`, `float`, etc
- Objects have properties (called _attributes_) and functions that define their abilities (called _methods_)
    - For example, can use the `__sizeof__` method to find the size of that object
- How to find out a summary of object / class attributes and methods?    
    - Use the `dir` method on an object to inspect all its attributes and methods
    - Access the `__dict__` attribute of a class to inspect all class attributes and methods
- Note that `int` objects have
    - Attributes such as `numerator`, `denominator`, `real`, `imag`
    - Methods such as `__abs__`, `__add__` etc

```python
a: int = 42
b: float = 1.01
print( a.__sizeof__(), b.__sizeof__() )
print( '*' * 42 )
print( [ ( attr, type( getattr( a, attr ) ) ) for attr in dir( a ) ] )
print( '*' * 42 )
int.__dict__
```

# Custom Classes
- Used to create custom datatypes
- Conventional for the class name to be in CamelCase not snake_case
- Methods can either operate on the object that called that method, or return a new one
    - All methods must take as their first argument the object that called that method
    - Pythonic to call that argument `self` although can use any other name too
- Special methods called _dunder_ methods or _magic_ methods
    - Usually called automatically and define a lot of the "under-the-hood" behavior of objects

# The `__init__` dunder
- `__init__` called **after** the object has been initially created
    - Meant to create / initialize all attributes of that object
    - Python allows any method to create / initialize object attributes
    - Best practice to use only the `__init__` method to do so
    - Creating object attributes outside of `__init__`  (updating attributes is okay) is bad idea
        - Makes code hard to read -- reader cannot quickly get a list of all attributes
        - Inconsistency: an object attribute created inside a method may not be present in all objects
            - Objects for whom that method is called will have that attribute
            - Objects for whom that method is not called will not have that attribute
            - Makes life hard for methods if different objects have different sets of attributes
            - `__init__` always gets called for each created object so ensures a consistent set of attributes
    - `__init__` can do typechecking to ensure object gets initialized to proper values
    - Recall that using _guard_ clauses are good way to write neat code that raises case-specific error messages

# Dunder Magics
- A beautiful instance of **polymorphism**: the same entity behaving differently as per context
    - The `__str__` dunder behaves differently for `int` vs `list` vs `dictionary` etc etc
- Can create one's own dunder methods but they wont be "magical"
    - Wont get called automatically but can be called explicitly as a regular object method
    - Not the best idea -- Python may create a new magic with the same name in a future release
    - Your code will not work as intended with future Python version as it would have shadowed the magic
- The magics `__add__`, `__radd__` etc are associated with arithmetic operations
- The magics `__lt__`, `__le__` etc are associated with comparison / ordering
    - If you define these magics for your class, then wont need to supply `key` when calling `sorted()`
- The magics `__str__`, `__repr__` are associated with printing and debugging
- The magics `__init__`, `__del__` are associated with creation and destruction of objects
    - `__init__` auto-called just **after** object is created
    - `__del__` auto-called just **before** object is destroyed (e.g. if someone used the `del` command)
        - This allows you to perform housekeeping e.g. if you defined a non-mutable datatype
    - Pay close attention to the "chronology": \<creation\> $\Rightarrow$ `__init__` $\Rightarrow$ \<lifecycle\> $\Rightarrow$ `__del__` $\Rightarrow$ \<deletion\>

# Object Attributes vs Class Attributes
- Defining (or assigning values to) attributes outside of any method creates class attributes
    - Meant to keep track of properties of the datatype itself rather than any object
    - **Warning**: Python allows accessing class attributes either via the object or the class
        - No issue if just reading the value
        - However, if class attribute is assigned a (new value) when accessed via object, it will cause a shadow
        - A new object attribute will be created (for only that object) with the same name as the class attribute
        - For that object, the object attribute will henceforth shadow the class attribute
        - Best practice to only access class attributes via class

# "Private" Attributes and Name Mangling
- Python has no notion of private or protected attributes or methods -- all attributes and methods are public
- However, Python "mangles" the names for attributes and methods with a special kind of name
    - These names are only accessible to methods of the class
    - Outsiders cannot access these directly as Python changes the name 
    - Mangled names start with a dunder but do not end with a dunder (can end with a sunder -- a single underscore)
    - However, the mangled name is no secret and can be still accessed "publicly" - inappropriate to do so though
    - **Note**: "sunder" is not a standard name but an "innovation" by the instructor
    - sunder `_` (single underscore), dunder `__` (double underscore), trunder `___` (triple underscore)
- **Pythonic Convention**: class attributes and methods whose name starts with a sunder are meant to be private
    - Polite request from whoever wrote that code to not access these directly
    - Not just a matter of politeness: it is likely that this is done since these are not meant to be accessed standalone
    - For example, the person who created a "private" variable may have intended special behavior
    - For example, if the private variable is changed, may require cascading changes to a class variable
    - If you forcibly change the private variable value but do not fix the class variable, will give wrong results

```python
class Fraction:
    # Class attributes
    prec = 2      # How many digits of precision to use when printing a Fraction in decimal form?
    pop = 0       # How many Fraction objects exist in the universe?
    
    def __init__( self, num = 0, den = 1 ):
        # guard clauses to check for type integrity
        if not type( num ) == int:
            raise TypeError( "numerator must be an integer" )
        if not type( den ) == int:
            raise TypeError( "denominator must be an integer" )
        if den == 0:
            raise ValueError( "denominator cannot be zero" )
        self.n = num
        self.d = den
        self.__p = f"my original value was {self}"
        if type( self ) == Fraction:
            Fraction.pop += 1

    def __frac__( s ):
        # Defining new object attributes allowed outside __init__ but not advisable
        s.new_attr = 42
        return f"{s.n / s.d:.{Fraction.prec}f}"

    # Return a printable representation of the object **value**
    def __str__( myself ):
        return f"{myself.n} / {myself.d}"

    # Return a description of how this object can be **created**
    # Mostly used in debugging
    def __repr__( s ):
        return f"Fraction( {s.n}, {s.d} )"

    def __add__( s, o ):
        if not type( o ) == Fraction:
            if type( o ) == int:
                o = Fraction( o )
            else:
                raise TypeError( "can only add Fraction to Fraction or int" )
        return Fraction( s.n * o.d + s.d * o.n, s.d * o.d )

    # Be careful -- if we return o + s, it will cause infinite recursion
    def __radd__( s, o ):
        return s + o

    def __sub__( myself, thyself ):
        return Fraction( myself.n * thyself.d - myself.d * thyself.n, myself.d * thyself.d )

    # Be careful -- if we return o - s, it will cause infinite recursion
    def __rsub__( s, o ):
        return -s + o

    def __neg__( s ):
        return Fraction( -s.n, s.d )

    def __iadd__( s, o ):
        return s + o
        
    # Use Euclid's algo to compute the GCD
    # Note that this method does not create a new object
    # Instead it modifies the object itself
    def reduce( s ):
        x, y = ( s.n, s.d )
        while y != 0:
            x, y = y, x % y
        s.n //= abs( x )
        s.d //= abs( x )
        # Class methods can access the "unmangled" names
        print( s.__p )

    def __del__( s ):
        if type( s ) == Fraction:
            Fraction.pop -= 1
```

```python
nlf = Fraction( 42, 2 )
print( nlf )
print( nlf.n, nlf.d ) # public attributes are okay to access directly
print( nlf._Fraction__p ) # impolite
nlf.reduce()
print( nlf )
```

```python
f = Fraction( 1, 4 )
print( Fraction.pop )
print( f )
f += f
print( f )
f.reduce()
print( f )
print( Fraction.pop )
del f
print( Fraction.pop )
```

```python
z = Fraction()
o = Fraction( num = 1 )
h = Fraction( 1, 2 )
t = Fraction( 1, 3 )

print( z.n, z.d )
print( z )
str( z )
print( o + z )
print( z - o )
print( h + 9 )
print( 10 + h )
print( (5 - h).__frac__() )
print( h.__frac__(), t.__frac__(), z.__frac__(), o.__frac__(), sep = ', ' )
Fraction.prec = 3
print( h.__frac__(), t.__frac__(), z.__frac__(), o.__frac__(), sep = ', ' )
# Uh oh -- we just shadowed a class attribute
t.prec = 6
print( t.prec )
Fraction.prec = 10
print( t.prec ) # No affect -- the object attribute has shadowed the class attribute
# Fortunately, the __frac__ method uses the class attribute and not the object attribute
# So printing will be unaffected by this shadow
print( h.__frac__(), t.__frac__(), z.__frac__(), o.__frac__(), sep = ', ' )
```
