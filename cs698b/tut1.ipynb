# NOPs in Python
- The body of `for`, `while` loop or any conditional clause `if`, `elif`, `else` cannot be empty
    - The body must contain at least one statement or expression
    - However, it is allowed to be effectively empty by including a _nop_ ( no operation ) statement called `pass`
    - Python communities usually use the `pass` statement to indicate that there is a _to-do_ here and code will be inserted later
- ### WARNING - Never use the %%timeit line / cell magic with commands that print stuff
    - The timeit magic may cause your code to run for an indeterminate number of iterations, maybe ( hundreds of ) thousands of iterations
    - If your code has print statements, it will cause those many lines of output to be printed
    - This may cause your **Jupyter server to freeze** and **maybe your system too** due to excessive memory usage
    - Better to use the perf_counter function that gives access to a global timer

```python
%%timeit
for i in range( 1000 ):
    pass
```

# Lists and Tuples
- Close cousins -- both represent ordered sequences of arbitrary objects
    - Python also supports unordered collections called _sets_ -- more on them later
- Lists are _mutable_, Tuples are _immutable_
- Lists are demarcated using square brackets, Tuples are demarcated using round parantheses
    - The third type of curly braces are used to demarcate _sets_ and _dictionaries_ -- more on them later
- Elements of lists / tuples are 0-indexed
    - Elements of lists / tuples can be any objects, even other tuples / lists
- Can do list arithmetic since lists can also be seen as forming a _Free Monoid_
- **Careful**: if creating a tuple with a single element, do not just say `( 3 )` or `( "hello" )`
    - Python will interpret this as an integer or a string, not a tuple containing an integer or tuple containing a string
    - Since math expressions also use round parentheses, Python does not intepret these as tuples
    - To force Python to interpret it as a tuple, must have a comma after the element e.g. `( 3, )` or `( "hello", )`
    - Recall, can use `_` for digit separation e.g. `10_00_000` but not `,`
    - If we say `10,00,000`, it will get interpreted as a tuple, even though there are no enclosing round parentheses
    - No such issue with lists -- `[ 3 ]` or `[ "hello" ]` will get interpreted as a list of a single element even without comma

```python
alpha = ( 'AA' ) # Will not work as expected
print( type( alpha ) )
alpha = ( 'AA', ) # Will work as expected
print( type( alpha ) )

# append will always "insert" at the end
# insert allows insertion at any location in the list
alpha = [ "AA" ]
for i in range( 26 ):
    alpha.insert( -1, chr( ord( 'A' ) + i ) )
print( alpha )

# pop is a method that removes by index
# if no index given, last index i.e. -1 removed as a default
# "popped" value returned
val = alpha.pop()
print( val, "was popped from", alpha )

# remove is a method that removes by value
# The first occurence of the value is removed
# Will cause an error if specified value is not present in the list
# Removed value is not returned
alpha.remove( 'P' )
print( alpha, "after removing 'P'" )

# del is a generic Python command to delete arbitrary objects
del alpha[ 0 ]
print( alpha, "after removing index 0" )
# Can use del to delete entire objects
# del alpha
# print( alpha ) # Will cause an error since the object itself has been deleted

# List arithmetic
list_ = list( range( 10 ) )
list_2 = list( range( 10, -1, -1 ) )
print( list_ )
print( list_ + list_2 )
print( 3 * list_ )
print( 2 * list_ * 3 )

# Creating lists from iteratables like range / zip
print( list( zip( list_, list_2, list_ ) ) )

# Zip will quietly stop zipping if one of its arguments runs out of elements
# Need to be careful since zip will quietly succeed without warning that it ran out of elements
print( list( zip( range( 10 ), range( 11, 16 ) ) ) )
```

# Lists and Tuples
- Lists support a lot of _methods_ ( not so many for tuples since they are immutable )
    - Methods are functions associated with a particular class
    - Can use the `in` operator to check if an object resides inside a list / tuple or not
        - To check for membership, value is compared, not memory address
    - Only objects of those classes can call those methods
    - Examples: `reverse()`, `append()`, `remove()`, `sort()` ( supports keys and reversing )

```python
# List / tuple nesting and indexing
# Lists and tuples can nest each other and contain arbitary objects
# Lists, tuples can contain heterogeneous objects
my_list = [ ( 1, 2, 3 ), "Hello World", [ 4, 5, 6 ] ]
my_tuple = ( my_list, "Good Morning", ( 1, 2, 3 ) )
print( my_list )
print( my_tuple )

# List membership using in operator and index() method
print( "Good Morning" in my_tuple )
print( "Hello World" in my_list )
sublist = [ 4, 5, 6 ]
# Membership check is by value not by address
print( sublist in my_list )
print( hex( id( sublist ) ), hex( id( my_list[ -1 ] ) ) )

# Unpacking / multiple assignment
( _, _, var ) = my_list[ 0 ]
print( var )
# No need for the enclosing parentheses
a, b, c = my_list
print( a, b, c )

# Splices can accept negative increments too
# Inverting a list using ::-1 shortcut
# Can alternatively use the reverse() method
print( my_list[ ::-1 ] )

# Lists support sorting -- can request sorting in "descending" or "reverse alphabetical" order by setting reverse = True
# If list is homogeneous and contains regular objects like numerals, strings etc then sorting will automatically do the expected thing
# If list is heterogenous, then need to specify how two objects must be compared for greater / lesser determination

list_ = list( range( 10 ) )
print( list_ )
list_.sort( reverse = True )
print( list_ )

list_2 = list( range( 10, -1, -1 ) )
new_list = list( zip( list_, list_2, list_ ) )
new_list[ 1 ] = "Hello World"
print( new_list )

# Cannot sort new_list since Python does not know how to compare a string to an integer
# Calling new_list.sort() will result in an error
# Must tell Python how to do this by defining a key function
# The key function must take each object within the list and return something that can be compared
# For every object in the list, the key function will get called and its outputs will get compared

def my_key( obj ):
    if type( obj[ 0 ] ) == type( "a" ):
        return ord( obj[ 0 ] )
    elif type( obj[ 0 ] ) == type( 0 ):
        return obj[ 0 ]

new_list.sort( key = my_key )
print( new_list )
```

# Lists and Tuples
- Python's storage model creates interesting effects with lists

```python
a = [ 1, 2, [ 1, 2, 3 ] ]
print( hex( id( a ) ) )
# Modifying a list does not change its storage address since it is a mutable object
a.append( 5 )
print( hex( id( a ) ) )

# Simple assignment does not create a new object, it simply creates a new name for the same object
b = a
print( a, b, hex( id( a ) ), hex( id( b ) ) )

# Modifications using the new name will affect the original object
b[ 0 ] = 99
print( a, b )

# If you really need a new object, need to use the in-built method copy()
c = a.copy()
print( a, b, c )

# This copy will behave independently in some cases
c[ 0 ] = 100
print( a, b, c )

# However, copy does not recursively copy nested lists
# The list embedded inside c is the same one as embedded inside a and b
c[ -2 ][0] = 10
print( a, b, c )

# To get a truly new copy, need to use the deepcopy method available from the copy module
```

# Comprehensions -- a popular form of Pythonic syntactic sugar
- Offer a loop, a conditional and an append into a neat sugary package
    - Turns out, this combination is often used in practice
    - Python anticipates and simplifies the most common use case
- Exist for lists, sets, dictionaries ( not for tuples )
    - Powerful, fabled, and can easily get difficult to "comprehend"
    - If you really want to do tuple comprehension, typecast the output of list comprehension to tuple
- Basic form for lists `[expr for value in collection if condition]`
    - Equivalent to running a loop over each `value` and appending `expr` to a list if `condition` is satisfied
    - List comprehension can always be avoided by using loops, conditionals instead -- matter of style
- Nested comprehensions are possible
    - Wonderful for designing complex exam questions
    - Two kinds of "nesting" possible
        - First kind is used to parse nested lists
        - Second kind is used to produce nested lists
    - Need to be very careful of order in nested comprehensions
- List comprehensions may offer a mild advantage over loops in terms of speed
    - Not a huge advantage due to optimization and because Python will "compile" code that is used again and again
    - Biggest advantage of list comprehension is in improved readability if used in moderation
    - Readability is lost if using 5 level nested list comprehension
    - Use `perf_counter()` to measure the timings instead of the risky `%%timeit` magic
    - Need to import via the `time` module using the `import` statement `from time import perf_counter`
    - Note that `perf_counter()` does not give the current time, it just gives access to a global timer
    - Values from `perf_counter()` are best used to calculate differences to obtain elapsed time

```python
# List comprehension demo
print( [ i ** 2 for i in range( 50 ) if i % 3 == 0 ] )

# Nested list comprehension to flatten a list of tuples
LoT = [ ( 1, 2, 3 ), ( 4, 5, 6 ), ( 7, 8, 9 ) ]
tmp = []
for tuple_ in LoT:
    for value in tuple_:
        if value % 2 == 0:
            tmp.append( value ** 2 )
print( tmp )
# The correct order in nested comprehension is the one the nested loops would have used
# Makes sense since after all, comprehensions are just syntactic sugar
tmp = [ value ** 2 for tuple_ in LoT for value in tuple_ if value % 2 == 0 ]
print( tmp )
# Note: order matters in nested list comprehensions
# The correct order may not be the intuitive one !
# List comprehension will give wrong result if wrong order is used
tmp = [ value ** 2 for value in tuple_ for tuple_ in LoT if value % 2 == 0 ]
print( tmp )

# Nested list comprehension can be used to produce a list of lists
print( [ [ i * j for i in range( 5 ) ] for j in range( 5 ) ] )

# Timed race between list comprehension and loop
from time import perf_counter

LoT = [ [ i * j for i in range( 2000 ) ] for j in range( 2000 ) ]

tic = perf_counter()
for tuple_ in LoT:
    for value in tuple_:
        if value % 2 == 0:
            tmp.append( value ** 2 )
toc = perf_counter()

print( toc - tic )

tic = perf_counter()
tmp = [ value ** 2 for tuple_ in LoT for value in tuple_ if value % 2 == 0 ]
toc = perf_counter()

print( toc - tic )
```

## Mini challenge 1
Write a program to read the text of an SMS or email and enumerate all numerical substrings present in that text, say mobile numbers, amounts etc

```python
text = input( '$ ' )
start_idx = [ i for i, ( a, b ) in enumerate( zip( '_' + text, text + '_' ) ) if not a.isdigit() and b.isdigit() ]
# print( start_idx )
end_idx = [ i for i, ( a, b ) in enumerate( zip( '_' + text, text + '_' ) ) if a.isdigit() and not b.isdigit() ]
# print( end_idx ) 
print( [ text[ s: e ] for ( s, e ) in zip( start_idx, end_idx ) ] )
```
