## Mini challenge 1 revisited
Write a program to read the text of an SMS or email and enumerate all numerical substrings present in that text, say mobile numbers, amounts etc

```python
text = input( '$ ' )
# Can compress both operations into a single operation that checks for digit-non-digit transistions
idx = [ i for i, ( a, b ) in enumerate( zip( '_' + text, text + '_' ) ) if a.isdigit() != b.isdigit() ]
print( [ text[ s: e ] for ( s, e ) in zip( idx[ ::2 ], idx[ 1::2 ] ) ] )
```

### List comprehension supports `else` clauses too
The general syntax needs to be flipped a bit
- Standard comprehension goes as `[ expr for value in collection if condition ]`
- It turns out we can specify the conditional before specifying the loop
    - Must specify an `else` clause in this case
        - `[ expr1 if condition else expr2 for value in collection ]`
    - Note: simply writing `[ expr1 if condition for value in collection ]` will cause an error
    - Can write multiple cases in this manner
        - `[ expr1 if condition1 else expr2 if condition2 else expr3 if condition3 else expr4 for value in collection ]`

```python
print( [ ( str( i ) + " is div2" ) if i % 2 == 0 else ( str( i ) + " is div3" ) if i % 3 == 0 else "None" for i in range( 10 ) ] )
```

## Mini challenge 2
Given an integer $n$, generate a list containing every number that can be constructed by selecting one or more digits from the number, while preserving the order in which the digits occur. This list must not contain any repeated entries ( e.g. if the number has repeated digits ). This list must be sorted in increasing order.
- The notion of _mask_ becomes very handy in certain programming problems
- Can generate all $l$-length masks simply by iterating over all integers from $0$ to $2 ^ l - 1$
- If only want non-empty masks, then iterate from $1$ to $2 ^ l - 1$
- Strings have a join method that allows a separator string to concatenate all elements in a list of strings
- The `ljust`, `rjust` string methods are helpful in right / left padding a strings with a pad character
- The `center` method will pad on both sides

```python
inp = 12256
subset_numbers = []
for i in range( 1, 2 ** 5 ):
    mask = str( bin( i ) )[ 2: ].rjust( 5, '0' )
    subset_numbers.append( int( ''.join( [ digit for ( digit, mask_digit ) in zip( str( inp ), mask ) if mask_digit == '1' ] ) ) )
print( sorted( set( subset_numbers ) ) )
```

## Mini challenge 3
Write a program that allows a human to play the game of **Bulls-and-Cows**. A secret 4 digit number is chosen with no digit repeating. To guess the number, the player must guess other 4 digit numbers that must also have no digit repeating. For each guess, the program responds with the number of bulls ( correct digits at correct positions ) and cows ( correct digit at wrong positions ). This classical game is an ancestor to commercial games like the NYT Wordle.
- Sets are a very common way to eliminate duplicates from a collection
    - Sets cannot have duplicates
    - Typecasting a list with duplicates into a set does not cause an error
    - Instead it quietly eliminates the duplicates
- Errors and exceptions are a way for a program to signal either improper input or improper operation
    - They can mostly be "caught" and "handled" -- will study them in detail later
- The use of **guard clauses** can greatly improve program readability, make debugging easier
    - Avoid having a long list of if-elif-else clauses checking for various invalid inputs
    - Instead, reject each form of invalid input using a separate conditional and accompanying error statement
    - Once all invalid cases have been handled, then proceed with regular processing ( with assurance of valid input )

```python
secret = "1234"
while True:
    guess = input( "Enter your guess: " )
    if len( guess ) != 4:
        raise Exception( "Invalid length of guess" )
    if len( guess ) != len( set( guess ) ):
        raise Exception( "Repeated digits in guess" )
    bulls = len( [ i for i, c in enumerate( secret ) if guess[ i ] == c ] )
    cows = len( set( secret ).intersection( set( guess ) ) ) - bulls
    print( f"{bulls} bulls, {cows} cows" )
```
