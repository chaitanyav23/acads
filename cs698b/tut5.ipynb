```python
string = "the good poor. big large fast. sluggish happy unhappy. old ancient hot. warm high tall. hard tough strong. sturdy bright clear. rich wealthy clean. tidy cat runs."
gold = {
    "swappable_count" : 24,
    "naive_signature" : "-010101010101010101010101--",
    "naive_watermark" : 5592405,
    "embed_Deebo_watermark" : "the good bad. big big fast. slow happy sad. old old hot. hot high high. hard hard strong. strong bright clear. wealthy rich tidy. tidy cat runs.",
}
```

```python
from synonyms import SYNONYM_GROUPS as sg # Aliasing
inv_idx = {}
for i, pair in enumerate( sg ):
    for j, word in enumerate( pair ):
        inv_idx[ word ] = i, j
```

```python
punc_tokens = string.split()
tokens = [ ( p_token if p_token[ -1 ].isalpha() else p_token[ :-1 ], '' if p_token[ -1 ].isalpha() else p_token[ -1 ] ) for p_token in punc_tokens ]
tokens
```

```python
my_swappable_count = sum( [ 1 for t in tokens if t[ 0 ] in inv_idx ] )
assert my_swappable_count == gold[ "swappable_count" ]
my_swappable_count
```

```python
my_naive_signature = ''.join( [ str( inv_idx[ t[ 0 ] ][ 1 ] ) if t[ 0 ] in inv_idx else '-' for t in tokens ] )
assert my_naive_signature == gold[ "naive_signature" ]
my_naive_signature
```

```python
my_naive_watermark = int( my_naive_signature.replace( '-', '' ), base = 2 )
assert my_naive_watermark == gold[ "naive_watermark" ]
my_naive_watermark
```

```python
str_len = len( tokens )
deebo_sig = bin( str_len )[ 2: ].rjust( my_swappable_count, '0' )
deebo_sig
```

```python
deebo_tokens = []
swap_idx = 0
for t in tokens:
    if t[ 0 ] not in inv_idx:
        deebo_tokens.append( t[ 0 ] + t[ 1 ] )
    else:
        deebo_tokens.append( sg[ inv_idx[ t[ 0 ] ][ 0 ] ][ int( deebo_sig[ swap_idx ] ) ] + t[ 1 ] )
        swap_idx += 1
my_embed_Deebo_watermark = ' '.join( deebo_tokens )
assert my_embed_Deebo_watermark == gold[ "embed_Deebo_watermark" ]
my_embed_Deebo_watermark
```
