# Bitwise operators
- Python supports the common binary operators `&`, `|`, `^`
- If used with Booleans, they represent the _and_, _or_, _exclusive-or (xor)_ operations
- If used with integers, they represent bitwise operators
- Python also supports bitshift operators `>>` and `<<`

## Mini challenge 4

Given an integer, check if it is a power of 2 and print its Boolean representation

```python
# Python supports integers with arbitary precision
# Can enter astronomically large integers and they will work just fine
n = int( input( "Enter an integer:" ) )

# Subtracting 1 from a decimal number like xxx10000 results in xxx09999
# Subtracting 1 from a binary number like xxx10000 results in xxx01111
# Subtracting 1 from an octal number like xxx10000 results in xxx07777
# Subtracting 1 from a hex number like xxx10000 results in xxx0FFFF
if not ( n & ( n - 1 ) ):
    print( "power-of-2" )
else:
    print( "not a power-of-2" )

str_n = ""
while n > 0:
    str_n += str( n & 1 )
    n = n >> 1

# Python's splicing is somewhat adaptive with respect to default values
# If only increment is specified and it is positive e.g. lst[ ::2 ], then it is interpreted as lst[ 0 : len( lst ) : 2 ]
# Note: lst[ 0 : -1 : 2 ] will not produce the same output -- why?
# If only increment is specified and it is negative e.g. lst[ ::-2 ], then it starts from len( a ) - 1 instead
print( str_n[ ::-1 ] )
```

## Mini challenge 5
Given a list of integers where every integer repeats twice except one, find the odd-one-out

```python
import random

# Create a list
int_list = list( range( 10 ) )
# Replicate each number
int_list *= 2
# Pop the last number making it the ( sole ) loner
int_list.pop()
# Shuffle the list randomly ( shuffles in-place -- no return )
random.shuffle( int_list )
print( int_list )

# XOR is associative and commutative
# a ^ b = b ^ a
# a ^ b ^ c = ( a ^ b ) ^ c = a ^ ( b ^ c )
# A serial XOR will cancel every duplicate and only leave the loner behind
accumulator = 0
for n in int_list:
    accumulator ^= n
print( accumulator )
```

# Dictionaries
- A mutable unordered collection datatype -- very popular form _key-value-pairs_
    - Keys in a dictionary must be immuatable -- values can be mutable
    - Can think of them as arrays but with _generalized_ non-integral indices
    - Order of keys in dictionary does not matter
    - A dictionary cannot have duplicate keys
- Must ensure a key is present in a dict before attempting recovery
    - Else will get `KeyError` ( why does Python itself use camelCase after telling everybody to use snake_case ??? )
        - Read PEP 8 in case you are curious
    - Similar to trying to access an array beyond its bounds
    - Can use the `get()` method -- syntactic sugar to check for existence else return a default value
    - Similarly, `setdefault()` method is syntactic sugar to set a key-value only if it does not exist
- Can use the `keys()`, `values()` and `items()` methods to obtain a list of keys, values and key-value pairs
    - This is a funny non-mutable list of type dict_keys. It is only meant to be used to iterate over
    - Can be creatively used to do a reverse membership check using the list index method
        - Need to typecast them into a list before using them though

```python
# Dictionary access, insert, item-wise modify, update ( with dictionary )

# Create a new dict
my_dict = {
    "hello" : "world",
    "1" : 0
}
print( my_dict )

# Updating a dict with an other dict will merge the two dicts
# Keys in the other dict not present in the calling dict will get added along with associated values
# Keys common to both dicts will have their values updated with value in the other dict
my_dict.update( { str( i ): i for i in range( 10 ) } )
print( my_dict )

# Pop is the safest way to remove elements from a dict
# Requires specifying the key to be popped -- make sure it is present else KeyError
my_dict.pop( '6' )
print( my_dict )

# Popitem is not a reliable way to implement item removal from a dict
# Its definition / implementation may change in future Python versions
# Code relying on a particular behavior of popitem may break in such cases
# Or be trapped using a specific ( outdated ) version of Python
my_dict.popitem()
print( my_dict )

# del, clear are also methods to remove dict items
# del will remove a specific key along with its value
# clear will remove all key-value pairs, leaving an empty dict

# Membership checks: key membership ( implicit ) vs value membership ( explicit ), (reverse) membership check
print( 7 in my_dict ) # Will not succeed since 7 is a value, not a key
print( 7 in my_dict.keys() ) # Checking membership in a dict is the same as checking membership in dict.keys()
print( 7 in my_dict.values() )

print( my_dict.keys(), type( my_dict.keys() ) )
print( my_dict.values(), type( my_dict.values() ) )
print( my_dict.items(), type( my_dict.items() ) )

# Dict values can be accessed by indexing with keys
# Dict does not support splicing since the keys need not be in order
# In a list, if index 0 and 3 are present, then 1 and 2 must also be present
# However, a dict may have key 0 and key 3 ( int keys ) but omit keys 1 and 2
# Makes operations such as splicing unsupported by dict
print( my_dict[ "7" ] )

# Reverse retrieval in a dict ( retrieving a key from a value )
# Will only recover the "first" instance
# If multiple instances of that value are present ( values can repeat in a dict, not keys )
# need to do list comprehension to identify all the indices and then retrieve corresponding keys
print( list( my_dict.keys() )[ list( my_dict.values() ).index( "world" ) ] )
```

# Dictionary Creation
- See Python docs for more details https://docs.python.org/3/library/stdtypes.html#dict
- Start with empty
- Using the `dict()` method in one of two ways
    - **Way 1**: keyword arguments of the form `dict( identifier1 = value1, identifier2 = value2, identifier3 = value3 )`
        - The identifiers ( variable names ) used become ( string-valued ) keys of the created dictionary
        - The identifiers may already exist or may be unused -- either way is fine since they are going to become strings anyway
        - This way can only create string-valued keys
    - **Way 2**: using iterables `dict( [ ( key1, value 1 ), ( key2, value2 ), ( key3, value3 ) ] )`
        - This way allows arbitrary types of keys to be used
- Using the `fromkeys()` method
    - Used to create a dictionary with every key having the same value
    - That same value is `None` by default but can be specified to be something else
- Note that `dict()` and `fromkeys() are a special type of function known as _constructors_
- Using (nested) dictionary comprehension --  works similarly to list comprehension, except that keys must be non-mutable and it creates a dict

```python
my_dict = {}

# Setting a value to a key can have two effects
# If the key already exists, its value gets updated
# The key does not exist, a new key-value pair is created
# To avoid this behavior, use the setdefault method
# setdefault will not update if key already exists
# setdefault will also return existing value as a side effect
my_dict[ "good" ] = "morning"
print( my_dict )

# The dict() function is a constructor and requires "keyword arguments"
my_dict = dict( a = 10, b = 20 )
print( my_dict )

my_var = 100
# Note: the first my_var is treated as a fresh identifier, meant to be converted into a string key
# The second my_var supplies the value for that key
my_dict = dict( my_var = my_var )
print( my_dict )

# Using a list of tuples of key-value pairs is also possible
my_dict = dict( [ ( "one", 1 ), ( 2, "two" ), ( "3", 3 ), ( 4, "4" ) ] )
print( my_dict )

# Note that invoking the function as `dict.fromkeys( my_list, value = 10 )` will cause an error
# This is because the function definition forces this to be a "positional parameter"
# See https://docs.python.org/3/library/stdtypes.html#dict for details
# A slash in the function definition enforces preceding parameters to be positional only
my_list = [ x ** 2 for x in range( 10 ) ]
print( dict.fromkeys( my_list, 10 ) )

# ( nested ) Dictionary comprehension
my_dict = { i : { chr( ord( 'A' ) + i ) : ord( 'A' ) + i } for i in range( 26 ) }
print( my_dict )
```
