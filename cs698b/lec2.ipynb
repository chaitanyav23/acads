# Getting started
- Python allows comments
    - Single-line comments are initiated using a # -- everything after it on that line is ignored by the interpreter
    - Python has no multi-line comments -- instead, the hack is to write a multi-line string that never gets used 🙂
- Python anticipates and simplifies the most common use case
    - print function will always end in a newline unless prevented from doing so by specifying some other `end` character
- The `input` function always yield strings -- they must be typecast as int, float etc if they are to be interpreted as other types

```python
# Taking input and print demo
inp = input( "Please enter a number" )
print( "The input was", inp )
print( type( inp ) )

# Printing multiple values in a single print statement
# print() accepts heterogenous values
inp = float( inp )
print( "The input was", inp )
print( type( inp ) )

# End character
print( "Hello", end = ':' )
print( "World" )

'''
This is a multiline comment
It is nothing but a string "constant" aka a string literal
Python does not require values to be typed or even be assigned to a variable
So this string literal will be read once and then be forgotten into the void
never to affect any future computation in any way, effectively what we want from a comment,
except that the interpreter does not even read single-line comments starting with a #
but an unoptimized interpreter will spend some time reading this multi-line comment
'''
```

# Primitive numeric datatypes: int, float
- Python computations rely on expressions that can be arithmetic, logical etc.
- Python's float is itself "double precision". Higher 128-bit precision available in numpy
    - Floating point imprecision ( e.g. 0.1 + 0.2 != 0.3 ) is a feature of machine implementation, not the Python language
    - Risky to compare floats for exact equality -- better to test for equality under a tolerance
    - For example, pytest explicitly encourages use of `approx` comparisons -- see judge code for lab

```python
# Simple expressions demo : arithmetic, modulus, (integer) division, exponentiation
# Spacing conventions

print(8**2)
print( 11 % 3 )
print( 6 // 4 )

# Type hints demo
a: int = 99
print( a, type( a ) )

# Python may ignore typehints demo
b: float = 99
print( b, type( b ) )

# Equivalence of 1 and 1.0
print( 4 == 4.00 )

# round, abs, bin(), hex() demos
print( round( 5.49 ), abs( -3 ), bin( 12 ), hex( 12 ) )
print( type( 0x1100 ) )

print( 0.1 + 0.2 == 0.3 )
print( 0.1 + 0.2 )

# Can only get binary representation for integers
print( bin( 0.1 ) )
```

# Primitive textual datatype: str
- `str` is a reserved keyword (used to obtain a textual representation of objects like numbers etc)
    - Improper / risky ( although allowed ) to use `str` as a variable name -- instead use `str_` that is safe yet looks like `str`
- Encapsulated by either double or single quotes
    - The two encapsulation techniques escape each other
- In-built function len() gives its length as an integer
- No explicit "character" datatype -- a unit-length string is a character
- Strings can mostly be interpreted as a sequence/array/ordered list of characters
    - Can do array-like access, splicing
    - Even string "arithmetic" is possible -- forms a structure called a _Free Monoid_
    - String addition works only if both objects are strings
    - String multiplication works only if one of the arguments is a string and the other an integer ( `str * int` and `int * str` both work )
- Python anticipates and simplifies the most common use case
    - When splicing, no need to specify both endpoints if either beginning at the start or ending at the finish

```python
# String encapsulation demo
str_ = 'Hello \' "World"'
print( str_ )

# len-type demo
print( len( str_ ) )
print( type( len( str_ ) ) )

# String indexing/slicing demo
str_[ -3: ]

# Ordinal function
char = 'c'
print( ord( 'A' ) <= ord( char ) <= ord( 'Z' ) )

# Python does not allow strings to be typecast to integer in general
# Unit length strings can be typecast if they can be interpreted as an integer, if base is correctly specified
print( int( '0' ) )
print( int( 'A' ) )
# For textual characters to be typecast, they must be interpretable as a hex character, if base is correctly specified
print( int( 'C', base = 16 ) )
print( int( 'G', base = 16 ) )

# String concatenation and replication -- float -> int, float/int -> str, str -> int?
print( str_ + str_ )
# 3 * str_ would have also worked -- will 3 * str_ * 3 work? why? why not?
print( str_ * 3 )
```

# Manipulating strings
- Instead of writing ( C-style ) `for( i = 0; i < 10; i ++ )`, we write `for i in range( 0, 10, 1 ):`
- Python anticipates and simplifies the most common use case
    - If starting point is 0 and increment is unity, then need only specify the post-destination ( i.e. destination + 1 )
    - Even further simplification -- if all we want is a single pass over an array, no need to even index
- The `enumerate` iterator allows taking a pass over the entire array but keeps track of their indices as well
- However, if we want fancy access that goes back and forth or has non-unit or uneven jumps, then these shortcuts do not apply
- f-strings ( formatted strings ) greatly simplify typesetting outputs
    - Expressions automatically get typecast into string before getting absorbed into the string

```python
# Traversing a string -- classical way using len()
for i in range( len( str_ ) ):
    print( str_[ i ], end = '_' )
print()

for c, i in enumerate( str_ ):
    print( c, i, end = ', ' )
print()

# Traversing a string -- Pythonic way
for c in str_:
    print( c, end = '_' )
print()

# f-strings demo
# The ( result of the ) numerical expression 10 * 3 automatically get 
print( f"Hello World can be written as {str_}, {10**3}" )
# If not using f-strings, it is our responsibility to typecast everything into string type
# Cannot concatenate a non-string to a string -- the following will give an error
# print( "Hello World" + 10 )
print( "Hello World" + str( 10 ) )
```
