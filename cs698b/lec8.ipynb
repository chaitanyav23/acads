# The under, dunder and trunder
- Three variables that store recently evaluated expressions
- `_` will store the most recently evaluated expression
- `__` will store the second most recently evaluated expression
- `___` will store the third most recently evaluated expression
- Only expressions not captured in a statement or variable are stored
- This is an iPython feature -- not available if writing a `.py` script
    - Try writing the following script in a `.py` file and running
      ```python
      5 + 3
      print( _ )
      ```
    - It will cause an error
      ```
      NameError: name '_' is not defined
      ```
- Can still use `_`, `__`, `___` as variable names in `.py` or `.ipynb`

# Python Modules
- Super easy to create a new module
- Even a standalone `.py` file can work as a module
    - Name of the file will become name of the module
- It can be `import`-ed -- we do this regularly in lab exercises
- Several ways to import modules
    - **General Import** : `import <module>`
        - Will import everything (functions, classes, variables) defined inside that module
        - A bit cumbersome to access those definitions e.g. `module.fun`, `module.VAR`, `module.CustomClass`
        - However, safe to use even if your own code uses names `fun`, `VAR`, `CustomClass`
    - **Specific Import** : `from <module> import <object>`
        - Will import only that specific object (function, class, variable) from that module
        - Easier to access that object -- no need to write `module.object` -- writing `object` enough
        - Riskier if your own code uses the same object name as it will cause a shadow situation
            - If your code object was defined earlier and import happened later, then your own object will get shadowed
            - If import happened earlier and your code object was defined later, then the module object will get shadowed
    - **Wildcard Import** : `from <module> import *`
        - Every object inside the module is imported specifically
        - Can be dangerous since your namespace will be full of hundreds or even thousands of names
        - You may not even know which all names are being shadowed -- use this only if you do not care about shadowing
    - **Aliased Import** : way to give nicknames to modules or objects therein -- can be used in general or specific import
        - `import <module> as <nick>` : all objects will be accessible as `nick.fun`, `nick.VAR`, `nick.CustomClass` etc
        - `from <module> import <object> as <nick>` : that object will have name `nick` -- shadow can still happen

# More Dunder Names: `__name__`
- Python declares standard dunder names with specific intent
- `__name__` will store the name of the top level execution environment
    - If the peice of code is being directly executed (e.g. running a `.py` file), `__name__ == "main"`
    - If the peice of code was imported into some other code (e.g. via `import`), `__name__ == <name-of-module>`
- The dunder name `__name__` is also found as an attribute in some objects
    - For example, for function objects, it will store the name of the function
- The idiomatic usage of the `__name__` dunder is to selectively execute code
    - Write code that is meant to be run if the `.py` file is executed by itself inside a `main()` function
    - Conditionally call the `main` function as follows
      ```python
      if __name__ == "main":
          main()
      ```
    - The `main()` function will not be executed if the file is imported as a module
- If executed standalone, `lec8.py` will produce the following output (notice that `__name__` is `"main"`)
  ```bash
    I can access the constant Good Morning FDE1
    My name is __main__
    My file is <path-to-file>/lec8.py
  ```
- However, if imported, the value of `__name__` changes to `lec8` (the name of the module)

# More Dunder Names: `__doc__`
- Yeilds the doctstring for a function, class etc.
- The docstring gets used at several places:
    - It gets attached to the `__doc__` attribute of an object (not the same as `__str__` or `__repr__`)
    - It gets returned whenver someone calls the `help()` function with that object
    - It gets returned if someone attempts introspection by using the `?` operator
        - `object?` will simply return the docstring for that object
        - `object??` will additionally return the code of that object (if written in Python)
        - If `??` does not return the code, there is a good chance the code was actually written in C
        - In such cases the function is just a "wrapper"
- Several styles to write docstrings: https://stackoverflow.com/questions/3898572/what-are-the-most-common-python-docstring-formats
