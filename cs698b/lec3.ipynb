# Unwanted variable: a most useful concept
- The underscore by itself is a valid variable name used internally by Python
    - Python will store the value of the last ( orphaned ) expression in `_`
    - Note that if the value of an expression gets assigned to a variable, then it does not get stored in `_`
- A Pythonic way to eliminate the need to create dummy variables just to catch unwanted values
    - Can be used to denote the loop variable in loops where the loop variable is dummy
- Python supports multiple assignments as a form of **syntactic sugar**
    - The underscore can catch (multiple) unwanted values in such multiple assignments
- The underscore also plays a role in variable names and literals
    - Can be used for digit grouping to improve readibility for long integers / floating point numbers
    - Conventionally used as a suffix if we really want to use reserved keywords as variable names
    - Conventionally used as a prefix to denote ( suggest althoguh not enforc e) a variable as private
- Double underscore ( aka "dunder" ) has other implications -- to be discussed later

```python
# Using _ to capture last-evaluated expression demo
b = 2 + 3
4 + 5
print( _ )
# The underscore can be used as a regular variable too
# However, uncommon and un-Pythonic to do so
_ = "Hello"
print( _ )

# Digit grouping demo
# The underscore can be used to better visualize long integers
salary = 10_00_000
print( salary, type( salary ) )
# Be careful to not use comma ( that will turn things into a tuple )
salary = 10,00,000
print( salary, type( salary ) )

# Unwanted variables in loops demo
for i in range( 2 ):
    print( "Hello World" )

for _ in range( 2 ):
    print( "Hello again" )

# Unwanted variables in (multiple) assignment demo
_, c = ( 2 + 3, 3 + 4 )
d = ( 2 + 3, 3 + 4 )
print( c, d )
# Can assignm multiple values to the same variable
# Not the best coding practice
# Reveals that multiple assignments take place left to right
_, _, e = ( 5, 6, 7 )
print( e, _ )
```

# Storage Model : Python vs C/C++/Java
- Major difference in how values and variables are related to each other
    - Sometimes one might say (somewhat inaccurately) that all Python variables are _references_
    - The in-built `id()` function gives the memory location of any object
    - The in-built `hex()` function gives a more compact hexadecimal representation of an integer
- Languages like C/C++/Java follow the _variable-value_ model that behaves like an _item-in-a-box_
    - Each variable is a memory location that can hold a value
    - Python instead follows the _name-object_ model that behaves like a _item-with-a-tag_
- Values are objects and variables are just names associate with those objects
    - Certain datatypes are immutable e.g. tuples, strings, integers
    - Any modifications to them creates a new object

```python
# Demo on implications of Python's name-object model
a = 10
b = 10
print( hex( id( a ) ), hex( id( b ) ), a, b )
b = 99
print( hex( id( a ) ), hex( id( b ) ), a, b )
c = ( 10, 20, 99 )
# There is only one object 10, only one object 99, etc even if they are used inside data structures like tuples
print( hex( id( c[ 0 ] ) ), hex( id( c[ 1 ] ) ), hex( id( c[ 2 ] ) ) )

# use of augmented operators +=, *=, /=, -=, %=
b: int = 10
print( hex( id ( b ) ), b )
b += 10 # Equivalent to b = b + 10
# As integers are immutable, changes to its value creates a new object
print( hex( id ( b ) ), b )
```

# Conditionals and Boolean type
- Unlike C / C++, Python explicitly recognizes a `bool` type that can take two values `True` and `False`
- Typecasting from non-Boolean to Boolean allowed -- usually follows C-convention of non-zero value being `True`, 0, empty string being `False` etc.
    - Concept of _truthy_ and _falsy_ values 🙂
    - If using inside conditionals, do not need to typecast to `bool`
    - Truthy values automatically interpreted as `True` and falsy values automatically interpreted as `False`
- Logical expressions involve comparison operators ( identical to C / C++ ) and logical operators ( `or`, `and`, `not` )
    - Continue to be careful about `=` vs `==`
    - The `==` operator can compare equality ( case sensitive ) between strings too
    - Best to avoid implicit comparisons between objects of different types e.g. `if "0" == 0`
    - Better to typecast before comparing
    - Risky to compare floats for equality ( instead, compare against a tolerance )

```python
# Boolean variables demo
# Implicit use of truthy / falsy values without typecasting
# Note: the integer 0 is a falsy value but the ( non-empty ) string "0" is a truthy value
if "0":
    print( "True" )
else:
    print( "False" )
    
# Comparison operators ==, !=, <, <=, >=, >
# Numerical, string, Boolean, types
# Hybrid comparisons

if "hello".lower() == "Hello".lower():
    print( "Equal" )
else:
    print( "Not" )

if "0" == True:
    print( "Equal" )
else:
    print( "Not" )

if "0" == str( 0 ):
    print( "Equal" )
else:
    print( "Not" )

print( bool( "0" ) )

if bool( "0" ) == True:
    print( "Equal" )
else:
    print( "Not" )
```

# Conditionals and Boolean type
- Conditionals usually create _blocks_ of code
    - Blocks boundaries must be declared using indentation -- take care to indent uniformly throughout
    - Can nest blocks -- Python provides three kinds of clauses `if`, `else` and `elif` ( short for else-if )
    - The `else` clause ( the _catch-all_ clause) is optional but if present, it must be the last clause and occur only once
    - Structure is `if` clause followed by zero or more `elif` clauses followed by zero or one `else` clause
    - `if` and `elif` must have a conditional statement but `else` is prohibited from having one ( as it is the _catch-all_ )
    - In an `if`-`elif`-`else` cascade, exactly one clause is ever activated
    - In an `if`-`elif` cascade ( no `else` clause ), it is possible that no clause is ever activated

# Loops
- Python supports standard loops `for`, `while`
- Key to mastering sophisticated loops is to design them using **Loop Invariants**
    - Think of these like proving a theorem by induction
    - A loop invariant is like a promise one iteration of the loop makes to the next one
    - The design of the promise must be linked to the problem being solved by the loop
    - An iteration ensures that if the previous promise held, then the next promise will hold
    - Thus, if the initial promise holds, then the loop must have done its job correctly
- `break` statement breaks out of a loop
    - Often used in conjunction with infinite loops
    - Some like this programming style, others hate it

```python
# Typing your name challenge
# Infinite loop with break
while True:
    name = input( "Input your name" )
    if name == "your name":
        break

# Example of loop invaraint
# Find the index of the largest element of a list that contains non-negative integers
# Loop invariant: after the i-th iteration, we have the index of the largest elements in the first i locations
array = [ 10, 9, 0, 6, 20, 2, 8, 5, 7 ]
# The base case is satisfied -- no iterations have executed so argmax is an illegal value
# What would happen if we initialize argmax / max to a legal value say 0?
argmax = -1
valmax = -1
for i, val in enumerate( array ):
    # What would happen if I use >= instead of >
    if val > valmax:
        valmax = val
        argmax = i
print( argmax, valmax )
```

# Loops
- Python supports `for`, `while` loops used in a variety of ways e.g with `zip`, `enumerate` etc
- `continue` statement helps is a way to write cleaner code
    - Can always avoid using continue by using conditionals instead but code can get messy
    - Good idea to use continue to give others a clear idea of your logic
- Loops can be nested -- can have loops / conditionals nested inside other loops / conditionals
- Note that `range` function accepts negative increments too

```python
# continue demo -- sum all digits occuring in a string
str_ = "Hello 123"
total = 0
for c in str_:
    # Clear indication to reader, for non-digits, the rest of the body of the loop is not executed
    if not c.isdigit():
        continue
    total += int( c )
print( total )

# Print decimal equivalent of a binary string
str_ = "12345"
val = 0
for i in range( len( str_ ) ):
    val = val * 10 + int( str_[ i ] )
print( val )
```
