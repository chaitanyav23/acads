# Tuple and Dictionary unpacking
- The `*` operator unpacks a tuple / list / dictionary
    - Can be useful if unpacking a tuple with an unknown number of elements
    - Can also be used to unpack a dictionary to create separate lists of keys and values
- The `**` operator unpacks a dictionary
    - Keys treated as identifiers ( variable names )
    - Mostly used in passing variable number of arguments to functions

```python
my_tuple = ( 1, 2, 3 )
print( "my_tuple", my_tuple )
print( "my_tuple upacked", *my_tuple, sep = ' # ' )
print( "list unpacked", *[ 4, 5, 6 ] )

d = { "name" : "Deebo", "roll" : 26007, "dept" : "AWSM" }
print( "dict", d )
print( "tuple unpacking with dict", *d )
print( "tuple unpacking the items", *d.items() )
print( "zip can work with more than two lists / tuples", list( zip( *d.items() ) ) )

# print( **d ) # Will not work since the print function does not take keyword arguments

def my_fun( name, roll, dept, hungry = True ):
    print( f"Hello {name} from {dept}", end = '' )
    if hungry:
        print( " , lets go to the mess!" )
    else:
        print()

# dict unpacking generates name-value pairs
my_fun( **d, hungry = False )
```

# Function arguments
- Python also allows functions with an unspecified number of arguments
    - Could have also been achieved by passing arguments in a list or a dictionary
    - Python offers syntactic sugar to not burden caller with packaging inputs in a tuple / list / dictionary
        - Use `*args` for positional arguments, `**kwargs` for keyword arguments
        - Do not confuse with `argc`, `*argv` in C -- despite similarity, `*args`, `**kwargs` are not pointers
        - This is how functions like `print()`, `zip()` are able to take an arbitrary number of arguments
- Python allows function arguments to take default values
- Python allows functions to be passed as arguments
    - Python functions are objects much like anything else
    - Allows useful in-built functions like `map()`, `filter()` and `sorted()` to work
